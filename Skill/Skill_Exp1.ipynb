```python
import pandas as pd

# File name
CSV_PATH = "automobile_dataset.csv"

# Read CSV file
df = pd.read_csv(CSV_PATH)

# 1. Display first 5 rows
print("FIRST 5 ROWS")
print(df.head())

# 2. Display dataset shape
print("\nDATASET SHAPE")
print(df.shape)

# 3. Display column names
print("\nCOLUMN NAMES")
print(df.columns)

# 4. Display data types
print("\nDATA TYPES")
print(df.dtypes)

# 5. Display basic information
print("\nDATASET INFORMATION")
df.info()

# 6. Display statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe(include='all'))

# 7. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 8. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# 9. Fuel Type distribution
print("\nFUEL TYPE DISTRIBUTION")
print(df["Fuel_Type"].value_counts())

# 10. Fuel Type percentage
print("\nFUEL TYPE PERCENTAGE")
print(df["Fuel_Type"].value_counts(normalize=True) * 100)

# 11. Transmission distribution
print("\nTRANSMISSION DISTRIBUTION")
print(df["Transmission"].value_counts())

# 12. Body Type distribution
print("\nBODY TYPE DISTRIBUTION")
print(df["Body_Type"].value_counts())

# 13. Average Selling Price
print("\nAVERAGE SELLING PRICE")
print(df["Selling_Price"].mean())

# 14. Maximum and Minimum Selling Price
print("\nMAXIMUM SELLING PRICE")
print(df["Selling_Price"].max())

print("\nMINIMUM SELLING PRICE")
print(df["Selling_Price"].min())
```
